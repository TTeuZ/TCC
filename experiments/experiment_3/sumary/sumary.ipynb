```python
from utils import print_confusion_matrix

import numpy as np
import json
import os
import re
```

```python
def get_cm(cm):
    temp = re.findall(r'\d+', cm)
    temp = np.array(temp, dtype=int).reshape(-1, 2)

    return temp
```

```python
def print_results(results):
    results.sort(key=lambda result: result["test"]["average"]["loss"])

    best_result = results[0]
    best_avg = best_result["test"]["average"]
    best_subsets = best_result["test"]["subsets"]

    avg_losses = [result["test"]["average"]["loss"] for result in results]
    avg_accuracy = [result["test"]["average"]["accuracy"] for result in results]

    print("-------------------------[ALL RESULTS]-------------------------")
    print(" - ".join(f"{loss:.4f}" for loss in avg_losses) + " [LOSS]")
    print(" - ".join(f"{acc:.4f}" for acc in avg_accuracy) + " [ACCURACY]")

    print("\n-------------------------[BEST RESULT]-------------------------")
    print(f"{best_avg['loss']:.4f} [LOSS]")
    print(f"{best_avg['accuracy']:.4f} [ACCURACY]")

    cm = get_cm(best_avg["cm"])

    print("")
    print_confusion_matrix(cm)

    print("\n-------------------------[ BY SUBSET ]-------------------------")
    subsets = [key for key in best_subsets.keys()]
    subsets.sort()
    for subset in subsets:
        cm = get_cm(best_subsets[subset]["cm"])

        print(f"{subset}: \t\t\t Confusion matrix")
        print(f"{best_subsets[subset]['loss']:.4f} [LOSS] \t\t\t {cm[1][1]} \t {cm[1][0]}")
        print(f"{best_subsets[subset]['accuracy']:.4f} [ACCURACY] \t\t {cm[0][1]} \t {cm[0][0]}")
        print("")
```

```python
RESULTS_PATH = "/home/tteuz/Desktop/TCC/experiment_3/_results"
results = [(file, json.load(open(f"{RESULTS_PATH}/{file}", "r"))) for file in os.listdir(RESULTS_PATH)]
```
