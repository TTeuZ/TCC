```python
import numpy as np
import cv2 as cv
import json
```

```python
def crop_rect(image, rect):
    center, size, angle = rect[0:3]
    height, width = image.shape[0:2]

    if int(angle) >= 45:
        angle = angle - 90
        temp = size
        size = (temp[1], temp[0])

    rotation_matrix = cv.getRotationMatrix2D(center, angle, 1)
    center, size = tuple(map(int, center)), tuple(map(int, size))

    image_rot = cv.warpAffine(image, rotation_matrix, (width, height))
    image_crop = cv.getRectSubPix(image_rot, size, center)

    return image_crop
```

```python
with open("/home/tteuz/Desktop/TCC/datasets/PKLot2.0/PKLot/ufpr05_spots.json", "r") as file:
    data = json.load(file)
```

```python
images_ids = set()

for annotation in data["annotations"]:
    images_ids.add(annotation["image_id"])

annotations = {image: [] for image in images_ids}

for annotation in data["annotations"]:
    annotations[annotation["image_id"]].append(annotation)
```

```python
annotation = annotations[0]
image_path = data["images"][0]["file_name"]
image = cv.imread(f"/home/tteuz/Desktop/TCC/datasets/PKLot2.0/PKLot/{image_path}")
```

```python
for crop in annotation:
    poly = np.array(crop["segmentation"])
    poly = np.reshape(poly, (-1, 2))

    rotaded_rect = cv.minAreaRect(poly)

    box = cv.boxPoints(rotaded_rect)
    box = np.int64(box)

    cv.drawContours(image, [box], 0, (0, 0, 255), 2)
```

```python
cv.imshow("Box", image)
cv.waitKey(0)

cv.destroyAllWindows()
```
