```python
import cv2 as cv
import json
```

```python
with open("/home/tteuz/Desktop/TCC/datasets/PKLot2.0/PKLot/pucpr_spots.json", "r") as file:
    pucpr = json.load(file)
```

```python
diff_images = set()

for annotation in pucpr["annotations"]:
    diff_images.add(annotation["image_id"])

pucpr_dict = {image: [] for image in diff_images}

for annotation in pucpr["annotations"]:
    pucpr_dict[annotation["image_id"]].append(annotation)
```

```python
annotation = pucpr_dict[0]
image_path = pucpr["images"][0]["file_name"]
```
