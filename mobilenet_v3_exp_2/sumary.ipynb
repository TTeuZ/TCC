```python
import json
import os
```

```python
RESULTS_PATH = "/home/tteuz/Desktop/TCC/mobilenet_v3_exp_2/_results"
results = [(file, json.load(open(f"{RESULTS_PATH}/{file}", "r"))) for file in os.listdir(RESULTS_PATH)]
```

```python
def print_results(results):
    file, best_result = results[0]

    losses = [result["test"]["loss"] for _, result in results]
    accuracies = [result["test"]["accuracy"] for _, result in results]

    avarage_loss = sum(losses) / len(losses)
    avarage_accuracy = sum(accuracies) / len(accuracies)

    print("First experiment results: ")

    print("Losses: ", end="")
    for loss in losses:
        print(f"{loss:.4f}", end=" ")
    print("", end="\n")

    print("Accuracies: ", end="")
    for accuracy in accuracies:
        print(f"{accuracy:.4f}", end=" ")
    print("\n")

    print("-------------------------------------------------------------------------------------------------------\n")

    print(f"Avarage loss: {avarage_loss:0.4f}")
    print(f"Avarage accuracy: {avarage_accuracy:0.4f}")

    print("\n-------------------------------------------------------------------------------------------------------\n")

    print(f"Best model loss: {best_result['test']['loss']:0.4f}")
    print(f"Best model accuracy: {best_result['test']['accuracy']:0.4f}")
    print(f"Best model file: {file}")
```

```python
first_experiment = []
second_experiment = []

for file, result in results:
    if result["dataset"]["train"] == "PKLotSegmented":
        first_experiment.append((file, result))
    else:
        second_experiment.append((file, result))

first_experiment.sort(key=lambda result: result[1]["test"]["loss"])
second_experiment.sort(key=lambda result: result[1]["test"]["loss"])
```

### First experiment

```python
print_results(first_experiment)
```

### Second experiment

```python
print_results(second_experiment)
```

### All experiments

```python
all_experiments = first_experiment + second_experiment
all_experiments.sort(key=lambda result: result[1]["test"]["loss"])
```

```python
print_results(all_experiments)
```
