```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np

import knn
import metrics
```

### Getting the train and test sets

```python
df_train = pd.read_csv("../features/train.csv")
df_test = pd.read_csv("../features/test.csv")
```

### Data normalization

```python
train_y = np.array(df_train.filter(["256"], axis=1))
train_x = np.array(df_train.drop(["dataset", "weather", "date", "hour", "parking_space", "256"], axis=1))

test_y = np.array(df_test.filter(["256"], axis=1))
test_x = np.array(df_test.drop(["dataset", "weather", "date", "hour", "parking_space", "256"], axis=1))
```

```python
scaler = MinMaxScaler()
scaler.fit(train_x)

normed_train = scaler.transform(train_x)
normed_test = scaler.transform(test_x)
```

### Model fit and prediction

```python
temp_train_x = normed_train[10450:10550]
temp_train_y = train_y[10450:10550]

temp_test_x = normed_test[103885:103935]
temp_test_y = test_y[103885:103935]
```

```python
model = knn.KNearestNeighbors()
model.fit(normed_train, train_y)

preds = model.predict(normed_test)
```

### Metrics

```python
print(f"Accuracy: {metrics.accuracy_score(temp_test_y, preds)}")
print("----------------------------------------------------------------------")

print("Confusion matrix:")
cm = metrics.confusion_matrix(temp_test_y, preds)
for j in reversed(range(2)):
    for i in reversed(range(2)):
        print(cm[i][j], end=' ')
    print('\n', end='')
print("----------------------------------------------------------------------")
```

### Metrics with sklearn

```python
print(f"Accuracy: {accuracy_score(temp_test_y, preds)}")
print("----------------------------------------------------------------------")

print("Confusion matrix:")
cm = confusion_matrix(temp_test_y, preds)
for j in reversed(range(2)):
    for i in reversed(range(2)):
        print(cm[i][j], end=' ')
    print('\n', end='')
print("----------------------------------------------------------------------")

print("Classification report:")
print(classification_report(temp_test_y, preds))
```
