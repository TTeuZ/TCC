```python
import sys, os; sys.path.append(os.path.join(os.path.dirname("avgs.ipynb"), '..'))

from tools.utils.utils import get_cm
from collections import defaultdict
import numpy as np
import json
```

```python
JSONS_PATH = "/home/tteuz/Desktop/TCC/cross_testing/_results/exp_f4e538f6-bb13-4b34-9460-106f9b4575df"

def get_experiments(path):
    jsons = [file for file in os.listdir(path) if ".json" in file]
    results = [json.load(open(f"{path}/{file}", "r")) for file in jsons]

    experiments = defaultdict(list)
    for result in results:
        experiments[result["dataset"]["train"]].append(result)

    return experiments
```

```python
experiments = get_experiments(JSONS_PATH)

result = experiments["CNRParkEXTSegmented"][0]
```

```python
# From the CM

cm = get_cm(result["test"]["average"]["cm"])
accuaracy = np.trace(cm) / np.sum(cm)
print(accuaracy)
```

```python
# Weighetd avg

enumerator, divider = 0, 0
for subset, values in result["test"]["subsets"].items():
    enumerator += values["accuracy"] * np.sum(get_cm(values["cm"]))
    divider += np.sum(get_cm(values["cm"]))

final_acc = enumerator / divider;

print(final_acc)
```
