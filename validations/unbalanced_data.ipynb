```python
import sys, os; sys.path.append(os.path.join(os.path.dirname("unbalanced_data.ipynb"), '..'))
from cross_testing.dataset.data_loader import data_loader
from tools.dataset.data_leveler import data_leveler
from collections import Counter
```

```python
DATASET_PATH = "/media/tteuz/ssd/datasets/PKLot2.0/PKLotSegmented"

ds_loader = data_loader()
train, val = ds_loader.load_dataset_by_split(DATASET_PATH, 0.7)
```

```python
print("------------------------------------------------")
print("train:")

class_0, class_1 = 0, 0
for subset in train.datasets:
    class_count = Counter(subset.targets)
    class_0 += class_count[0]
    class_1 += class_count[1]

total = class_0 + class_1
perc_class_0 = (class_0 / total) * 100
perc_class_1 = (class_1 / total) * 100

print(f"Total Class 0: {class_0} - {perc_class_0:.2f} | Total class 1: {class_1} - {perc_class_1:.2f}")

print("------------------------------------------------")
print("result:")

ds_leveler = data_leveler()
temp = ds_leveler.flatten_dataset(train)

class_0, class_1 = 0, 0
for subset in temp.datasets:
    class_count = Counter(subset.targets)
    class_0 += class_count[0]
    class_1 += class_count[1]

total = class_0 + class_1
perc_class_0 = (class_0 / total) * 100
perc_class_1 = (class_1 / total) * 100

print(f"Total Class 0: {class_0} - {perc_class_0:.2f} | Total class 1: {class_1} - {perc_class_1:.2f}")
```

```python
print("------------------------------------------------")
print("val")

class_0, class_1 = 0, 0
for subset in val.datasets:
    class_count = Counter(subset.targets)
    class_0 += class_count[0]
    class_1 += class_count[1]

total = class_0 + class_1
perc_class_0 = (class_0 / total) * 100
perc_class_1 = (class_1 / total) * 100

print(f"Total Class 0: {class_0} - {perc_class_0:.2f} | Total class 1: {class_1} - {perc_class_1:.2f}")

print("------------------------------------------------")
print("result:")

ds_leveler = data_leveler()
temp = ds_leveler.flatten_dataset(val)

class_0, class_1 = 0, 0
for subset in temp.datasets:
    class_count = Counter(subset.targets)
    class_0 += class_count[0]
    class_1 += class_count[1]

total = class_0 + class_1
perc_class_0 = (class_0 / total) * 100
perc_class_1 = (class_1 / total) * 100

print(f"Total Class 0: {class_0} - {perc_class_0:.2f} | Total class 1: {class_1} - {perc_class_1:.2f}")
```
